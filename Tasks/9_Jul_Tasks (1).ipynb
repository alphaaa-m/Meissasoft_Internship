###Problem:
Given an array, reverse the elements without using a second array.

```python
arr=list(map(int,input().split()))

l=0
r=len(arr)-1
while l<r:
  arr[l],arr[r]=arr[r],arr[l]
  l+=1
  r-=1

print(*arr)
```

```python
def binary_search(arr, x,l=0,h=(len(arr))-1):
  if h>=l:
    m=int((h+l)/2)
    if x==arr[m]:
      return m
    elif x>arr[m]:
      return binary_search(arr,x,m,h)
    elif x<arr[m]:
      return binary_search(arr,x,l,m)
    else:
      return -1
```

###Problem:
Print the first N numbers in the Fibonacci series using recursion.

```python
def fibonacci_sequence(n):
    sequence = [0,1]
    while len(sequence) < n:
        next_term = sequence[-1] + sequence[-2]
        sequence.append(next_term)
    return sequence


fibonacci_sequence(10)
```

###Problem:
Find the shortest path from a source vertex to all other vertices in a graph.

```python
graph = {
    'A': {'B': 1, 'C': 4, 'D':6},
    'B': {'C': 2, 'D':3},
    'C': {'D':4},
    'D':{}
}
```

```python
def dijkstra(graph, start):
    dist = {node: float('inf') for node in graph}
    dist[start] = 0
    visited = set()

    while len(visited) < len(graph):
        curr = None
        smallest_dist = float('inf')
        for node in graph:
            if node not in visited and dist[node] < smallest_dist:
                smallest_dist = dist[node]
                curr = node

        if curr is None:
            break

        for neighbor, weight in graph[curr].items():
            if dist[curr] + weight < dist[neighbor]:
                dist[neighbor] = dist[curr] + weight

        visited.add(curr)

    return dist
```

```python
dijkstra(graph,'A')
```

###Problem:
Given weights and values of items, find the maximum value that can be obtained within a given weight capacity.

```python
weights = [1, 3, 4, 5]
values = [1, 4, 5, 7]
capacity = 7
```

```python
def knapsack(weights, values, capacity):
    n = len(values)

    table = []
    for i in range(n + 1):
      row = []
      for w in range(capacity + 1):
        row.append(0)
      table.append(row)


    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            weight = weights[i - 1]
            value = values[i - 1]

            if weight <= w:
              taken = value + table[i - 1][w - weight]
              skipped = table[i - 1][w]
              table[i][w] = max(taken, skipped)
            else:
              table[i][w] = table[i - 1][w]

    return table[n][capacity]
```

```python
knapsack(weights, values, capacity)
```

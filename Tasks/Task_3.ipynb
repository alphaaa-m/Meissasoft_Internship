###Question 2: Write a Python program to find the factorial of a number.

```python
n = int(input("Number: "))
def fact(n):
  if n == 0 and n == 1:
    return 1
  elif n<0:
    print('no factorial')
  else:
    for i in range(1,n):
      n = n*i
    return n

fact(n)
print(f"The factorial of {n} is {fact(n)}")
```

###Question 3: Write a Python program to check if a number is prime.

```python
num=int(input("Number: "))

def prime(num):
  if num<=1:
    print('not prime')
  else:
    for i in range (1,num):
      if num%i==0:
        print('not prime')
        break
      else:
        print('prime')
        break

prime(num)
```

###Question 5: Write a Python program to remove duplicates from a list.

```python
import numpy as np
x = (np.random.randint(1, 20, 10))
print(x)
new = []
for i in x:
  if i not in new:
    new.append(i)
print("removed duplicates:", new)
```

### Question 1: Write a Python program to check if a string is a palindrome.

```python
x= input()
print('original:',x)
new=''
for i in x:
  new=i+new
print('reversed:',new)
if x==new:
  print('palindrome')
else:
  print('not palindrome')
```

```python
def dynamicArray(n, queries):

    arr = [[] for _ in range(n)]

    lastAnswer = 0

    results = []
    for query in queries:
        query_type = query[0]
        x = query[1]
        y = query[2]

        idx = (x ^ lastAnswer) % n

        if query_type == 1:
            arr[idx].append(y)
        else:
            element_idx = y % len(arr[idx])

            lastAnswer = arr[idx][element_idx]

            results.append(lastAnswer)

    return results

if __name__ == '__main__':

    first_multiple_input = input().rstrip().split()

    n = int(first_multiple_input[0])

    q = int(first_multiple_input[1])

    queries = []

    for _ in range(q):
        queries.append(list(map(int, input().rstrip().split())))

    result = dynamicArray(n, queries)

    print('\n'.join(map(str, result)))
```

```python
class Node:
    def __init__(self, info):
        self.info = info
        self.left = None
        self.right = None
        self.level = None

    def __str__(self):
        return str(self.info)

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def create(self, val):
        if self.root == None:
            self.root = Node(val)
        else:
            current = self.root

            while True:
                if val < current.info:
                    if current.left:
                        current = current.left
                    else:
                        current.left = Node(val)
                        break
                elif val > current.info:
                    if current.right:
                        current = current.right
                    else:
                        current.right = Node(val)
                        break
                else:
                    break

def preOrder(root):
    if root is None:
        return
    print(root.info, end=' ')
    preOrder(root.left)
    preOrder(root.right)


tree = BinarySearchTree()
t = int(input())

arr = list(map(int, input().split()))

for i in range(t):
    tree.create(arr[i])

preOrder(tree.root)
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

def equalStacks(h1, h2, h3):
    sum1 = sum(h1)
    sum2 = sum(h2)
    sum3 = sum(h3)

    while h1 and h2 and h3:
        if sum1 == sum2 == sum3:
            return sum1
        if sum1 >= sum2 and sum1 >= sum3:
            sum1 -= h1.pop(0)
        elif sum2 >= sum1 and sum2 >= sum3:
            sum2 -= h2.pop(0)
        else:
            sum3 -= h3.pop(0)

    return 0

if __name__ == '__main__':
    first_multiple_input = input().rstrip().split()

    n1 = int(first_multiple_input[0])
    n2 = int(first_multiple_input[1])
    n3 = int(first_multiple_input[2])

    h1 = list(map(int, input().rstrip().split()))
    h2 = list(map(int, input().rstrip().split()))
    h3 = list(map(int, input().rstrip().split()))

    result = equalStacks(h1, h2, h3)

    print(str(result))
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

def twoStacks(maxSum, a, b):
    count = 0
    sum_a = 0
    i = 0
    j = 0

    while i < len(a) and sum_a + a[i] <= maxSum:
        sum_a += a[i]
        i += 1
    count = i

    while j < len(b) and i >= 0:
        sum_a += b[j]
        j += 1
        while sum_a > maxSum and i > 0:
            i -= 1
            sum_a -= a[i]
        if sum_a <= maxSum:
            count = max(count, i + j)

    return count

if __name__ == '__main__':
    g = int(input().strip())

    for g_itr in range(g):
        first_multiple_input = input().rstrip().split()

        n = int(first_multiple_input[0])

        m = int(first_multiple_input[1])

        maxSum = int(first_multiple_input[2])

        a = list(map(int, input().rstrip().split()))

        b = list(map(int, input().rstrip().split()))

        result = twoStacks(maxSum, a, b)

        print(str(result))
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

def largestRectangle(h):
    stack = []
    max_area = 0
    i = 0
    while i < len(h):
        if not stack or h[i] >= h[stack[-1]]:
            stack.append(i)
            i += 1
        else:
            top = stack.pop()
            area = h[top] * (i if not stack else i - stack[-1] - 1)
            max_area = max(max_area, area)

    while stack:
        top = stack.pop()
        area = h[top] * (i if not stack else i - stack[-1] - 1)
        max_area = max(max_area, area)

    return max_area


if __name__ == '__main__':
    n = int(input().strip())

    h = list(map(int, input().rstrip().split()))

    result = largestRectangle(h)

    print(str(result))
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys
import heapq

def cookies(k, A):
    heapq.heapify(A)
    operations = 0

    while len(A) > 1 and A[0] < k:
        first = heapq.heappop(A)
        second = heapq.heappop(A)
        new_cookie = first + 2 * second
        heapq.heappush(A, new_cookie)
        operations += 1

    if A[0] >= k:
        return operations
    else:
        return -1


if __name__ == '__main__':
    first_multiple_input = input().rstrip().split()

    n = int(first_multiple_input[0])

    k = int(first_multiple_input[1])

    A = list(map(int, input().rstrip().split()))

    result = cookies(k, A)

    print(str(result))
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys
import heapq

def minimumAverage(customers):
    customers.sort()

    n = len(customers)
    current_time = 0
    total_waiting_time = 0
    customer_index = 0
    min_heap = []

    while customer_index < n or min_heap:
        while customer_index < n and customers[customer_index][0] <= current_time:
            heapq.heappush(min_heap, (customers[customer_index][1], customers[customer_index][0]))
            customer_index += 1

        if not min_heap:
            current_time = customers[customer_index][0]
            continue

        processing_time, arrival_time = heapq.heappop(min_heap)

        waiting_time = current_time - arrival_time + processing_time
        total_waiting_time += waiting_time

        current_time += processing_time

    return total_waiting_time // n


if __name__ == '__main__':
    n = int(input().strip())

    customers = []

    for _ in range(n):
        customers.append(list(map(int, input().rstrip().split())))

    result = minimumAverage(customers)

    print(str(result))
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys
from collections import deque

def minimumMoves(grid, startX, startY, goalX, goalY):
    n = len(grid)
    queue = deque([(startX, startY, 0)])
    visited = set([(startX, startY)])

    while queue:
        x, y, moves = queue.popleft()

        if x == goalX and y == goalY:
            return moves

        moves_list = [(0, 1), (0, -1), (1, 0), (-1, 0)]

        for dx, dy in moves_list:
            nx, ny = x + dx, y + dy
            while 0 <= nx < n and 0 <= ny < n and grid[nx][ny] != 'X':
                if (nx, ny) not in visited:
                    visited.add((nx, ny))
                    queue.append((nx, ny, moves + 1))
                nx += dx
                ny += dy

    return -1


if __name__ == '__main__':
    n = int(input().strip())

    grid = []

    for _ in range(n):
        grid_item = input()
        grid.append(grid_item)

    first_multiple_input = input().rstrip().split()

    startX = int(first_multiple_input[0])

    startY = int(first_multiple_input[1])

    goalX = int(first_multiple_input[2])

    goalY = int(first_multiple_input[3])

    result = minimumMoves(grid, startX, startY, goalX, goalY)

    print(str(result))
```

Task 42
- Components in a graph

```python
#!/bin/python3

import math
import os
import random
import re
import sys

class DSU:
    def __init__(self, n):
        self.parent = list(range(n + 1))
        self.size = [1] * (n + 1)

    def find(self, i):
        if self.parent[i] == i:
            return i
        self.parent[i] = self.find(self.parent[i])
        return self.parent[i]

    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)
        if root_i != root_j:
            if self.size[root_i] < self.size[root_j]:
                root_i, root_j = root_j, root_i
            self.parent[root_j] = root_i
            self.size[root_i] += self.size[root_j]
            return True
        return False

def componentsInGraph(gb):
    n = len(gb) * 2
    dsu = DSU(n)

    for u, v in gb:
        dsu.union(u, v)

    component_sizes = {}
    for i in range(1, n + 1):
        root = dsu.find(i)
        if dsu.size[root] > 1:
            component_sizes[root] = dsu.size[root]

    if not component_sizes:
        return [0, 0]

    min_size = min(component_sizes.values())
    max_size = max(component_sizes.values())

    return [min_size, max_size]


if __name__ == '__main__':
    n = int(input().strip())

    gb = []

    for _ in range(n):
        gb.append(list(map(int, input().rstrip().split())))

    result = componentsInGraph(gb)

    print(' '.join(map(str, result)))
```

Task 57:
- Down to zero

```python
import math
import random
import re
import sys
from collections import deque

def downToZero(n):
    q = deque([(n, 0)])
    visited = {n}

    while q:
        current_num, steps = q.popleft()

        if current_num == 0:
            return steps

        next_num = current_num - 1
        if next_num >= 0 and next_num not in visited:
            visited.add(next_num)
            q.append((next_num, steps + 1))

        for i in range(int(math.sqrt(current_num)), 1, -1):
            if current_num % i == 0:
                next_num = current_num // i
                if next_num not in visited:
                    visited.add(next_num)
                    q.append((next_num, steps + 1))
                break

if __name__ == '__main__':
    q = int(input().strip())

    for q_itr in range(q):
        n = int(input().strip())

        result = downToZero(n)

        print(str(result))
```

Task 72:
- The Power Sum

```python
def powerSum(X, N):
    max_base = int(X ** (1/N))

    memo = {}

    def count_ways(sum_left, base):
        if (sum_left, base) in memo:
            return memo[(sum_left, base)]

        if sum_left == 0:
            return 1
        if sum_left < 0 or base > max_base:
            return 0

        power = base ** N

        if power > sum_left:
            return 0

        include = count_ways(sum_left - power, base + 1)
        exclude = count_ways(sum_left, base + 1)

        total = include + exclude
        memo[(sum_left, base)] = total
        return total

    return count_ways(X, 1)

X = int(input())
N = int(input())
print(powerSum(X, N))
```

```python
import itertools

def alternate(s):
    unique_chars = list(set(s))
    max_length = 0

    if len(unique_chars) < 2:
        return 0

    for char1, char2 in itertools.combinations(unique_chars, 2):
        temp_s_list = []
        for char_s in s:
            if char_s == char1 or char_s == char2:
                temp_s_list.append(char_s)

        temp_s = "".join(temp_s_list)

        is_alternating = True
        if len(temp_s) > 1:
            for i in range(1, len(temp_s)):
                if temp_s[i] == temp_s[i-1]:
                    is_alternating = False
                    break

        if is_alternating and len(temp_s) > 1:
            max_length = max(max_length, len(temp_s))

    return max_length

alternate('beabeefeab')
```

```python
def kruskals(g_nodes, g_from, g_to, g_weight):

    class UnionFind:
        def __init__(self, n):
            self.parent = list(range(n + 1))
            self.rank = [0] * (n + 1)

        def find(self, i):
            if self.parent[i] == i:
                return i
            self.parent[i] = self.find(self.parent[i])
            return self.parent[i]

        def union(self, i, j):
            root_i = self.find(i)
            root_j = self.find(j)

            if root_i != root_j:
                if self.rank[root_i] < self.rank[root_j]:
                    self.parent[root_i] = root_j
                elif self.rank[root_j] < self.rank[root_i]:
                    self.parent[root_j] = root_i
                else:
                    self.parent[root_j] = root_i
                    self.rank[root_i] += 1
                return True
            return False
    edges = []
    for i in range(len(g_from)):
        u = g_from[i]
        v = g_to[i]
        wt = g_weight[i]
        edges.append((wt, u, v, u + v + wt))
    uf = UnionFind(g_nodes)

    mst_weight = 0
    edges_in_mst_count = 0

    edges.sort()

    for wt, u, v, _ in edges:
        if uf.union(u, v):
            mst_weight += wt
            edges_in_mst_count += 1

            if edges_in_mst_count == g_nodes - 1:
                break

    return mst_weight

g_nodes, g_edges= map(int,input().split())

g_from_list = []
g_to_list = []
g_weight_list = []

for _ in range(g_edges):
  u, v, w = map(int, input().split())
  g_from_list.append(u)
  g_to_list.append(v)
  g_weight_list.append(w)


print(kruskals(g_nodes, g_from_list, g_to_list, g_weight_list))
```

- Prim's (MST): special subtree

```python
import heapq

def prims(n, edges, start_node):
    adj = [[] for _ in range(n + 1)]
    for u, v, w in edges:
        adj[u].append((w, v))
        adj[v].append((w, u))

    visited = [False] * (n + 1)

    min_heap = [(0, start_node)]
    heapq.heapify(min_heap)

    minimum_cost = 0

    while min_heap:
        weight, u = heapq.heappop(min_heap)

        if visited[u]:
            continue

        visited[u] = True
        minimum_cost += weight

        for neighbor_weight, v in adj[u]:
            if not visited[v]:
                heapq.heappush(min_heap, (neighbor_weight, v))

    return minimum_cost

n, m = map(int, input().split())

edges = []
for _ in range(m):
    u, v, w = map(int, input().split())
    edges.append((u, v, w))

start_node = int(input())

result = prims(n, edges, start_node)
print(result)
```

```python
import collections

def quickestWayUp(ladders, snakes):
    board_map = list(range(101))

    for start, end in ladders:
        board_map[start] = end

    for start, end in snakes:
        board_map[start] = end

    dist = [-1] * 101

    queue = collections.deque()

    dist[1] = 0
    queue.append(1)

    while queue:
        current_square = queue.popleft()

        if current_square == 100:
            return dist[100]

        for die_roll in range(1, 7):
            next_pos_after_roll = current_square + die_roll

            if next_pos_after_roll > 100:
                continue

            final_landing_pos = board_map[next_pos_after_roll]

            if dist[final_landing_pos] == -1:
                dist[final_landing_pos] = dist[current_square] + 1
                queue.append(final_landing_pos)

    return -1


t = int(input())

for _ in range(t):
    n = int(input())
    current_ladders = []
    for _ in range(n):
        sl, el = map(int, input().split())
        current_ladders.append((sl, el))

    m = int(input())
    current_snakes = []
    for _ in range(m):
        ss, es = map(int, input().split())
        current_snakes.append((ss, es))

    result = quickestWayUp(current_ladders, current_snakes)
    print(result)
```

- Dijkstra: Shortest Reach 2

```python
import heapq
import sys
from collections import defaultdict

def shortestReach(n, edges, s):
    adj = [[] for _ in range(n + 1)]
    edge_map = {}

    for u, v, w in edges:
        if (u, v) not in edge_map or w < edge_map[(u, v)]:
            edge_map[(u, v)] = w
            edge_map[(v, u)] = w

    for (u, v), w in edge_map.items():
        adj[u].append((v, w))

    dist = [float('inf')] * (n + 1)
    dist[s] = 0

    visited = [False] * (n + 1)
    heap = [(0, s)]

    while heap:
        d, u = heapq.heappop(heap)

        if visited[u]:
            continue
        visited[u] = True

        for v, w in adj[u]:
            if not visited[v] and dist[v] > d + w:
                dist[v] = d + w
                heapq.heappush(heap, (dist[v], v))

    return [dist[i] if dist[i] != float('inf') else -1 for i in range(1, n + 1) if i != s]


if __name__ == '__main__':
    t = int(input())
    for _ in range(t):
        n, m = map(int, input().split())
        edges = [tuple(map(int, input().split())) for _ in range(m)]
        s = int(input())
        print(*shortestReach(n, edges, s))
```

```python
def largestPermutation(k, arr):
    n = len(arr)
    if k >= n - 1:
        arr.sort(reverse=True)
        return arr
    pos = [0] * (n + 1)
    for i in range(n):
        pos[arr[i]] = i
    for i in range(n):
        if k == 0:
            break
        target_value = n - i

        if arr[i] == target_value:
            continue

        current_idx_of_target_value = pos[target_value]

        value_at_current_pos = arr[i]

        arr[i], arr[current_idx_of_target_value] = arr[current_idx_of_target_value], arr[i]
        pos[target_value] = i
        pos[value_at_current_pos] = current_idx_of_target_value
        k -= 1

    return arr



n, k = map(int,input().split())
arr = list(map(int, input().split()))
print(*largestPermutation(k, arr))
```

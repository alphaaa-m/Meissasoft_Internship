##Big O Notation



- Big O notation is a mathematical tool used to describe the efficiency of an algorithm, indicating how its performance (time or space) scales with the size of the input. It focuses on the worst-case scenario and helps compare algorithms by abstracting away constant factors and lower-order terms. Essentially, it provides an upper bound on the growth rate of an algorithm's resource consumption.

- Imagine you're baking cookies. Big O Notation is like telling you how much longer it will take if you decide to bake a whole lot more cookies. It doesn't tell you the exact minutes, just how the time "grows" with more cookies.

##Time Complexity

- Time complexity quantifies the amount of time an algorithm takes to run as a function of the input size, expressed in terms of the number of operations performed. It doesn't measure actual execution time but rather the growth rate of operations, allowing for an abstract comparison of algorithmic efficiency. For example, an O(n) algorithm's operations grow linearly with input size, while an O(n^2) algorithm's operations grow quadratically.
- This is about how much time a computer program needs to finish its job. If you give it more stuff to work on, does it take a little more time, or a LOT more time?

##Space Complexity

- Space complexity measures the amount of auxiliary memory an algorithm requires to run as a function of the input size, excluding the space for the input itself. It helps evaluate how memory usage grows with increasing input, distinguishing between algorithms that use a constant amount of memory (O(1)) versus those whose memory footprint scales with the input size (O(n) or O(n^2)). This is crucial for optimizing memory-constrained applications.
- This is about how much memory a computer program needs to do its job. Think of it as how much counter space your cookie ingredients take up. Does it need more space if you bake more cookies, or does it always fit in the same spot?

## Summary

- Big O notation is the language we use to articulate the time complexity and space complexity of algorithms. Time complexity describes how an algorithm's execution time scales with input size, while space complexity describes how its memory usage scales. Together, these concepts provide a comprehensive framework for analyzing and comparing the efficiency of different algorithms, guiding developers in choosing the most performant solutions for given constraints.
- So, Big O Notation is the secret code we use to talk about how much time and space a computer program will need as it gets more stuff to do. It helps us pick the fastest and least "memory-hungry" programs!

##Linked List

```python
class Node():
  def __init__(self,data):
    self.data=data
    self.next=None

class LinkedList():
  def __init__(self):
    self.head=None
  def append(self, data):
    new_node=Node(data)
    if self.head==None:
      self.head=new_node
      return

    current=self.head
    while current.next:
      current=current.next
    current.next=new_node

  def display(self):
      current=self.head
      while current:
        print(current.data, end='-->')
        current = current.next
      print(None)

  def len(self):
    count=0
    current=self.head
    while current:
      count+=1
      current=current.next
    return count
```

```python
ll=LinkedList()
```

```python
ll.append(10)
ll.append(2)
ll.append(13)
```

```python
ll.display()
```

```python
ll.append(1)
```

```python
ll.display()
```

```python
ll.len()
```

```python
ll.append([2,3])
```

```python
ll.display()
```

```python
ll.len()
```

## Hash Tables

Dictionaries are hash tables in python

key, value pair is involved in dict
hash function is used to pair key and values

hash function basically converts a key in to some specific number that assigns a value at some index

```python
def hash(key, size):
  hash=0
  for char in key:
    hash+=ord(char)        #ord is unicode of characters
  return hash%size
```

```python
hash('c',10)
```

```python
hash('a',10)
```

```python
hash('c',100)
```

```python
#backend of dictionary

class HashTable():
  def __init__(self):
    self.size=10
    self.table=[None for i in range(self.size)]
  def hash(self,key):
    hash=0
    for char in key:
      hash+=ord(char)
    return hash%self.size
  def add(self,key,value):
    index=self.hash(key)
    self.table[index]=value
  def get_val(self,key):
    h=self.hash(key)
    return self.table[h]
```

##Tree

In trees there are `root nodes`, `nodes` and `leaf nodes`
(grandparents, parents, childern and so on)




```python
class TreeNode():
  def __init__ (self, data):
    self.data=data
    self.children=[]
    self.parent=None

  def add_child(self,child):
    child.parent=self
    self.children.append(child)

  def print_tree(self, level=0):
    indent='   ' * level
    print(f'{indent} - {self.data}')
    for child in self.children:
      child.print_tree(level+1)
```

```python
root=TreeNode('Vehicles')
```

```python
bike=TreeNode('Bike')
car=TreeNode('Car')

root.add_child(bike)
root.add_child(car)
```

## Graph

two main things are nodes and edges  
types of graphs
- directed graphs
- undirected graphs
- weighted graphs
- unweighted graphs

```python
class Graph():
  def __init__(self):
    self.graph={}
  def add_vertex(self,vertex):
    if vertex not in self.graph:
      self.graph[vertex]=[]
  def add_edge(self, u,v,weight=None, directed=False):       #directed=True means one way False means two way
    self.add_vertex(u)
    self.add_vertex(v)
    self.graph[u].append((v,weight))
    if not directed:
      self.graph[v].append((u,weight))
  def get_neighbors(self,vertex):
    return self.graph.get(vertex, [])
  def print_graph(self):
    for vertex, neighbors in self.graph.items():
      neighbor_strings = [f"{neighbor}({weight})" for neighbor, weight in neighbors]
      print(f"{vertex}: {', '.join(neighbor_strings)}")
```

```python
#unidirected unweighted graphs
g1=Graph()
g1.add_edge('ali', 'bilal')
g1.add_edge('bilal', 'ahmad')
g1.add_edge('ahmad', 'nasir')
g1.add_edge('nasir', 'bilal')
g1.add_vertex('ali')

g1.print_graph()
```

```python
g1.get_neighbors('ahmad')
```

```python
g1.get_neighbors('bilal')
```

```python
#directed weighted graph
g2=Graph()
g2.add_edge('A', 'B', 4, True)
g2.add_edge('A', 'C', 2, True)
g2.add_edge('B', 'D', 1, True)
g2.add_edge('C', 'D', 5, True)
g2.add_edge('C', 'E', 3, True)
g2.add_edge('D', 'E', 1, True)
g2.add_edge('E', 'B', 2, True)

g2.print_graph()
```

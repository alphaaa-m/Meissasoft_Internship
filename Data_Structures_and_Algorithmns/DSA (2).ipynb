##Linear Search

Linear search is a sequential searching algorithm where we start from one end and check every element of the list until the desired element is found. It is the simplest searching algorithm.  
- Time Complexity: O(n)
- Space Complexity: O(1)

```python
def LinearSearch(arr,n,x):
  for i in range(0,n):
    if (arr[i]==x):
      return i
  return -1                   #no x is found in the array
```

```python
import numpy as np
```

```python
arr=np.random.randint(0,100,20)
n=len(arr)
x=np.random.randint(0,100)
LS=LinearSearch(arr,n,x)
print(arr)
print(x)
if (LS==-1):
  print(f'{x} is not in the array')
else:
  print(f'{x} is at index {LS}')
```

##Binary Search

Binary Search is a searching algorithm for finding an element's position in a sorted array.

In this approach, the element is always searched in the middle of a portion of an array.

Binary Search Algorithm can be implemented in two ways which are discussed below.

- Iterative Method
- Recursive Method (divide and conquer approach)

- Time Complexity: O(log n)
- Space Complexity: O(1)

_Iterative Method_

```python
arr=np.random.randint(0,100,50)
arr.sort()
n=len(arr)
x=np.random.randint(0,100)
print(f'Array is {arr}, \nSize of array is {n}, \nElement to find is {x}')
```

```python
def B_S_Iterative(arr,n,x):
  low = 0
  high =n-1
  while low<=high:
    mid = int((low+high)/2)
    if x==arr[mid]:
      return mid
    elif x>arr[mid]:
      low=mid+1
    else:
      high=mid-1
  return -1
```

```python
BS1=B_S_Iterative(arr,n,x)
if (BS1==-1):
  print(f'{x} is not in the array')
else:
  print(f'{x} is at index {BS1}')
```

_Recursive Method_

```python
def B_S_Recursive(arr,x, low, high):
  if high >= low:
    mid = int((low+high)/2)
    if x==arr[mid]:
      return mid
    elif x>arr[mid]:
      return B_S_Recursive(arr,x,mid+1,high)
    else:
      return B_S_Recursive(arr,x,low,mid-1)
  else:
    return -1
```

```python
BS2=B_S_Recursive(arr,x,0,n-1)
if (BS2==-1):
  print(f'{x} is not in the array')
else:
  print(f'{x} is at index {BS2}')
```

## Bubble Sort

Bubble sort is a sorting algorithm that compares two adjacent elements and swaps them until they are in the intended order.

- Time Complexity: O($n^2$)
- Space Complexity: O(1)

```python
arr=np.random.randint(0,100,50)
n=len(arr)
print(f'Array is {arr}, \nSize of array is {n}')
```

```python
def BubbleSort(arr):
  for i in range(n):
    for j in range(0, n-i-1):
      if arr[j]>arr[j+1]:
        x=arr[j]
        arr[j]=arr[j+1]
        arr[j+1]=x
```

```python
BS=BubbleSort(arr)
print(f'Sorted array in ascending order is \n{arr}')
```

##Selection Sort

Selection Sort is a comparison-based sorting algorithm. It sorts an array by repeatedly selecting the smallest (or largest) element from the unsorted portion and swapping it with the first unsorted element.

- Time Complexity: O($n^2$)
- Space Complexity: O(1)

```python
arr=np.random.randint(0,100,50)
n=len(arr)
print(f'Array is {arr}, \nSize of array is {n}')
```

```python
def SelectionSort(array):
  for x in range(n):
    min=x
    for i in range(x+1,n):
      if array[i]<array[min]:
        min=i
    array[x],array[min]=array[min],array[x]
```

```python
SelectionSort(arr)
print(f'Sorted array in ascending order is \n{arr}')
```

##Insertion Sort

Insertion sort is a simple sorting algorithm that works by iteratively inserting each element of an unsorted list into its correct position in a sorted portion of the list.

- Time Complexity: O($n^2$)
- Space Complexity: O(1)

```python
arr=np.random.randint(0,100,50)
n=len(arr)
print(f'Array is {arr}, \nSize of array is {n}')
```

```python
def InsertionSort(array):
  for i in range(1,n):
    key=array[i]
    j=i-1
    while j>=0  and key < array[j]:
      array[j+1]=array[j]
      j=j-1
      array[j+1]=key
```

```python
I_S=InsertionSort(arr)
print(f'Sorted array in ascending order is \n{arr}')
```

## Merge Sort

Merge Sort is one of the most popular sorting algorithms that is based on the principle of Divide and Conquer Algorithm.

A problem is divided into multiple sub-problems. Each sub-problem is solved individually. Finally, sub-problems are combined to form the final solution.

- Time Complexity: O(nlogn)
- Space Complexity: O(n)

```python
def Merge(L,R):
  merged=[]
  i=0
  j=0
  while i <len(L) and j <len(R):
    if L[i]<=R[j]:
      merged.append(L[i])
      i+=1
    else:
      merged.append(R[j])
      j+=1

  while i<len(L):
    merged.append(L[i])
    i+=1
  while j<len(R):
    merged.append(R[j])
    j+=1
  return merged

def MergeSort(array):
  if len(array)<=1:
    return array
  mid=len(array)//2
  L=array[:mid]
  R=array[mid:]
  L=MergeSort(L)
  R=MergeSort(R)
  return Merge(L,R)
```

## Quick Sort

Quicksort is a sorting algorithm based on the divide and conquer approach where, an array is divided into subarrays by selecting a pivot element (element selected from the array)  and then splitting the rest of the list into two parts: those smaller than the pivot and those larger than the pivot.


- Time Complexity: O(nlogn)
- Space Complexity: O(n)

```python
arr=np.random.randint(0,100,50)
n=len(arr)
print(f'Array is {arr}, \nSize of array is {n}')
```

```python
def QuickSort(array):
  if len(array)<=1:
    return array
  pivot=array[-1]
  less=[]
  equal=[]
  greater=[]
  for x in array:
    if x<pivot:
      less.append(x)
    elif x==pivot:
      equal.append(x)
    else:
      greater.append(x)
  return QuickSort(less)+equal+QuickSort(greater)
```

```python
Q_S=QuickSort(arr)
print(f'Sorted array in ascending order is \n{Q_S}')
```

Time Complexity: O(n)

```python
n=int(input())
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
print(f"Factorial of {n} is {factorial(n)}")
```

## Two Pointers Technique

Two pointers is really an easy and effective technique that is typically used for Two Sum in Sorted Arrays, Closest Two Sum, Three Sum, Four Sum, Trapping Rain Water and many other popular interview questions. Given a sorted array arr (sorted in ascending order) and a target, find if there exists any pair of elements (arr[i], arr[j]) such that their sum is equal to the target.

Time Complexity: O($n^2$)

Space Complexity: O(1)

```python
def two_sum(arr, target):
    n = len(arr)

    for i in range(n):

        for j in range(i + 1, n):

            if arr[i] + arr[j] == target:
                return True

    return False

arr=np.random.randint(-100,100,20)
print(arr)
target = int(input('Target: '))

if two_sum(arr, target)==True:
    print("true")
else:
    print("false")
```

## Activity Selection Problem

Given n activities with start times in start[ ] and finish times in finish[ ], find the maximum number of activities a single person can perform without overlap. A person can only do one activity at a time.

```python
import pandas as pd
import random

def generate_activities(num, min_start=0, max_duration=10, max_finish=24):
    data = []
    for i in range(num):
        start = random.randint(min_start, max_finish - 1)
        finish = random.randint(start + 1, min(max_finish, start + max_duration))
        data.append({'Name': f'Act_{i+1}', 'Start': start, 'Finish': finish})
    return pd.DataFrame(data)

def select_activities(df):
    if df.empty:
        return pd.DataFrame(columns=['Name', 'Start', 'Finish'])
    df = df.sort_values('Finish').reset_index(drop=True)
    selected = [dict(df.loc[0])]
    last_finish = df['Finish'][0]
    for i in range(1, len(df)):
        if df['Start'][i] >= last_finish:
            selected.append(dict(df.loc[i]))
            last_finish = df['Finish'][i]
    return pd.DataFrame(selected)

if __name__ == "__main__":
    print("Activity Selection")
    print("\nRandom Activities")
    random_df = generate_activities(10, 0, 5, 8)
    print("\nInput:")
    print(random_df.to_string(index=False))
    selected_random = select_activities(random_df)
    print("\nSelected:")
    print(selected_random.to_string(index=False) if not selected_random.empty else "No activities selected.")
    print(f"\nTotal: {len(selected_random)} activities")
```

## Huffman Encoding

Huffman coding is a lossless data compression algorithm. The idea is to assign variable-length codes to input characters, lengths of the assigned codes are based on the frequencies of corresponding characters.
The variable-length codes assigned to input characters are Prefix Codes, means the codes (bit sequences) are assigned in such a way that the code assigned to one character is not the prefix of code assigned to any other character. This is how Huffman Coding makes sure that there is no ambiguity when decoding the generated bitstream.

```python
text=str(input())
```

```python
from collections import Counter
freq=Counter(text)
freq
```

```python
freq_list = [[char, freq[char]] for char in freq]
print("Frequency List:", freq_list)
```

```python
nodes = [[item[1], [item[0]]] for item in freq_list]

while len(nodes) > 1:
    nodes = sorted(nodes, key=lambda x: x[0])

    left = nodes.pop(0)
    right = nodes.pop(0)

    merged_freq = left[0] + right[0]
    merged_tree = [left, right]

    nodes.append([merged_freq, merged_tree])
```

```python
codes = {}

def assign_codes(node, code=""):
    if isinstance(node[1], list) and isinstance(node[1][0], str):
        codes[node[1][0]] = code
    else:
        assign_codes(node[1][0], code + "0")
        assign_codes(node[1][1], code + "1")
assign_codes(nodes[0])

print("Huffman Codes:", codes)
```

```python
encoded = ""
for char in text:
    encoded += codes[char]

print("Encoded text:", encoded)
```

## Breadth-First Search (BFS)

BFS is a fundamental graph traversal algorithm that explores a graph level by level. It's excellent for finding the shortest path in unweighted graphs.

- Time complexity: O(V + E), where V is the number of nodes and E is the number of edges.

- Space complexity: O(V).

```python
graph = {
    0: [1, 2],
    1: [2],
    2: [3],
    3: [1, 2]
}

def BFS(graph_map, start_node):

    seen = set()

    places_to_visit = []

    places_to_visit.append(start_node)

    seen.add(start_node)

    print("BFS Traversal Order:")

    while places_to_visit:

        current_place = places_to_visit.pop(0)

        print(str(current_place) + " ")

        for next_place in graph_map[current_place]:
            if next_place not in seen:
                seen.add(next_place)
                places_to_visit.append(next_place)

BFS(graph, 0)
```

## Dijkstra’s Algorithm

Dijkstra’s Algorithm is a shortest path algorithm used to find the minimum distance from a starting point (called the source) to all other nodes in a graph — like finding the shortest route from your house to every other place in a city.

```python
graph = {
    'A': {'B': 1, 'C': 4, 'D':6},
    'B': {'C': 2, 'D':3},
    'C': {'D':4},
    'D':{}
}
```

```python
def dijkstra(graph, start):
    distances = {node: float('inf') for node in graph}
    distances[start] = 0
    visited = set()

    while len(visited) < len(graph):
        current_node = None
        current_distance = float('inf')

        for node in graph:
            if node not in visited and distances[node] < current_distance:
                current_distance = distances[node]
                current_node = node

        if current_node is None:
            break
        for neighbor, weight in graph[current_node].items():
            new_distance = distances[current_node] + weight
            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance

        visited.add(current_node)

    return distances
```

```python
dijkstra(graph, 'A')
```

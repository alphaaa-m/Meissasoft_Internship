> **Instance method** – belongs to an object of the class and can access or modify instance attributes. It always takes self as the first parameter, referring to the current object.

###Encapsulation

The Art of Data Hiding and Bundling

- Bundling data and methods into a single unit (class), controlling access to internal state.

- Data Protection, Information Hiding

```python
class BankAccount:
  def __init__(self):
    self._balance= 0.0               #protected attribute
  @property
  def balance(self):
    return self._balance

  def deposit(self, amount):
    if amount<=0:
      raise ValueError('Amount should be positive')
    self._balance+=amount

  def withdraw(self, amount):
    if amount<=0:
      raise ValueError('Amount should be positive')
    if amount>self._balance:
      raise ValueError('Insufficient balance')
    self._balance-=amount
```

```python
acc=BankAccount()
```

```python
acc.balance
```

```python
acc.balance=20     #we can not directly change the balance of the account
```

```python
print(f'initial balance is {acc.balance}')
acc.deposit(500)
print(f'new balance after depositing is {acc.balance}')
acc.withdraw(100)
print(f'new balance after withdrawing is {acc.balance}')
```

we forced the user to use deposit and withdraw in order to change the balance instead of directly changing it,
the user don't know what happened in the class while depositing and withdrawing but it gets a desired output

```python
acc.balance
```

###Abstraction
Simplifying Complexity
- Focusing on essential features while hiding irrelevant details; providing a simplified view.
- Complexity Reduction, Improved Efficiency

```python
class EmailService:
  def _connect(self):                       #protected method
    print ('Connecting to server...')
  def _authenticate(self):                  #protected method
    print ('Authenticating...')
  def send(self):                           #public method
    self._connect
    self._authenticate
    print ('Sending email...')
    self._disconnect
  def _disconnect(self):                    #protected method
    print ('Disconnecting from the server...')
```

```python
email=EmailService()
```

```python
email.send()          #user can send email without using irrelevant details
```

###Inheritance
Building on Existing Foundations

- involves creating new classes (subclass or derived class) based on existing class (base class or super class)
- acquires properties and behaviors from an existing class, code reusability

```python
#super class
class Vehicle:
  def __init__(self, brand, model, year):
    self.brand=brand
    self.model=model
    self.year=year
  def start(self):
    print('Starting the vehicle')
  def stop(self):
    print('Stopping the vehicle')

#sub class
class Car(Vehicle):
  def __init__(self, brand, model, year, wheels, doors):   #have attributes of the parent class plus its own
    super().__init__(brand, model, year)      #atttributes of the superclass
    self.wheels=wheels
    self.doors=doors

#sub class
class Bike(Vehicle):
  def __init__(self, brand, model, year, wheels, color):
    super().__init__(brand, model, year)
    self.wheels=wheels
    self.color=color
```

###Polymorphism
Poly means many and morphs means forms _ Many Forms, One Interface
- allows objects of different classes be used as a common class or interface, enabling one interface to handle different behaviors.
- Code Reusability, Flexibility, Extensibility

```python
class Car():
  def __init__(self, brand, model, year, wheels, doors):
    self.brand=brand
    self.model=model
    self.year=year
    self.wheels=wheels
    self.doors=doors

  def start_car(self):
    print('Starting the car')
  def stop_car(self):
    print('Stopping the car')

class Bike():
  def __init__(self, brand, model, year, color):
    self.brand=brand
    self.model=model
    self.year=year
    self.color=color

  def start_bike(self):
    print('Starting the bike')
  def stop_bike(self):
    print('Stopping the bike')
```

The \_\_str\_\_() function controls what should be returned when the class object is represented as a string.

If the \_\_str\_\_() function is not set, the string representation of the object is returned:
